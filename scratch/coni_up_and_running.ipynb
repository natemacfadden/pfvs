# Benchmark

```python
import time
import tracemalloc
from tqdm.auto import tqdm
from ZpAndreas import *
```

```python
import smarter_conipfv
```

Settings

```python
do_old    = True
monitor_memory = False
requested_N_pvecs  = 20
ellipsoid_dilation = 200
```

```python
# containers for various data
h11 = []
h21 = []

num_pfv_old = []
num_pfv_fp  = []
num_pfv_gu  = []

time_old = []
time_fp  = []
time_gu  = []

memory_old = []
memory_fp  = []
memory_gu  = []

# misc data
first_run = True

for coni in tqdm(conis):
    # prep the data for PFV runs
    # --------------------------
    cy   = coni.dual_triangulation().cy()
    data = cydata.CYData.from_cy(cy, coni_curve=coni.conifold_charge())
    Qmin = cy.h11()+cy.h21()+4
    Qmax = cy.h11()+cy.h21()+4

    print(cy.h11(),cy.h21(),end=' ',flush=True)
    if cy.h11() >= 6:
        print('skipping h11=11+')
        continue

    # get p-vectors
    ps = Zp.pvecs(data, requested_N_pts=requested_N_pvecs, backend='gurobi')

    # save general data
    # -----------------
    h11.append(cy.h11())
    h21.append(cy.h21())

    gvs = cy.compute_gvs(max_deg=10).coo

    # iterate over p-vectors
    for p in ps:
        # old method
        # ----------
        if (cy.h11() <= 9) and do_old:
            t0 = time.time()
            if monitor_memory:
                tracemalloc.start()
    
            # construct PFVs
            ps_old, Ms_old, Ks_old = coniZp(cy,
                   min_points=1, pvecs=[[0] + p],
                   max_denominator=500,basis_matrix=data.cob,gvs=gvs,verbosity=0,maximum_box_size=1e5,doW0Cut=False,ncf=2,print_progress=False,tadpole_add=1,max_K=250)
            Ks_old = Ks_old.astype(int)
            Ms_old = Ms_old.astype(int)
            
            # trim PFVs with M[0] <= 12
            bad_M0 = Ms_old[:,0] <= 12
            bad_M0 = bad_M0 | (-np.sum(Ks_old*Ms_old,axis=1)!=Qmax)
            Ks_old = Ks_old[~bad_M0].tolist()
            Ms_old = Ms_old[~bad_M0].tolist()

            if monitor_memory:
                current, peak = tracemalloc.get_traced_memory()
                tracemalloc.stop()
                memory_old.append(peak)
            t1 = time.time()
    
            # save the data
            num_pfv_old.append(len(Ks_old))
            time_old.append(t1-t0)

        
        # new (FP) method
        # ---------------
        if first_run:
            Zp.coniZpM(
                data=data,
                ps=[ps[0]],
                Qmin=Qmax,
                Qmax=Qmax,
                ellipsoid_dilation=ellipsoid_dilation,
                max_N_pfvs=100_000_000,
                verbosity=0)
            first_run = False
    
        t0 = time.time()
        if monitor_memory:
            tracemalloc.start()

        # construct PFVs
        Ks_fp, Ms_fp = Zp.coniZpM(
            data=data,
            ps=[p],
            Qmin=Qmax,
            Qmax=Qmax,
            ellipsoid_dilation=ellipsoid_dilation,
            max_N_pfvs=100_000_000,
            verbosity=0)

        Ks_fp = Ks_fp.tolist()
        Ms_fp = Ms_fp.tolist()

        if monitor_memory:
            current, peak = tracemalloc.get_traced_memory()
            tracemalloc.stop()
            memory_fp.append(peak)
        t1 = time.time()

        # save the data
        num_pfv_fp.append(len(Ks_fp))
        time_fp.append(t1-t0)

        
        # new (Gurobi) method
        # -------------------
        t0 = time.time()
        if monitor_memory:
            tracemalloc.start()

        # get ellipsoid info
        mat, Z, Binter = Zp.coniMellipsoid(np.concatenate([[0],p]),data)

        # construct PFVs
        daa()
        Ks, Ms = smarter_conipfv.coniMan(
            Binter=Binter,
            Z=Z,
            Qmin=Qmax,
            Qmax=Qmax,
            M0min=13,
            Kscalebounds = 100,
            cbounds=100,
            alphabounds=1000,
            K0max=100,
            Nsol=1000,
            timelimit=60,
            verbosity=0)

        Ks_gu = []
        Ms_gu = []
        for K,M in zip(Ks,Ms):
            test = validation.PFV(data, K, M, silent=1)
            try:
                if test.check_all():
                    Ks_gu.append(K)
                    Ms_gu.append(M)
            except Exception as e:
                pass#print(e)
        if len(Ks_new):
            Ks_gu = np.vstack(Ks_gu).tolist()
            Ms_gu = np.vstack(Ms_gu).tolist()

        if monitor_memory:
            current, peak = tracemalloc.get_traced_memory()
            tracemalloc.stop()
            memory_gu.append(peak)
        t1 = time.time()

        # save the data
        num_pfv_gu.append(len(Ks_gu))
        time_gu.append(t1-t0)
```
